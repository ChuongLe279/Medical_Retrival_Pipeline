```python
import chromadb
from chromadb.utils.embedding_functions import SentenceTransformerEmbeddingFunction

# 1. Initialize persistent Chroma client
client = chromadb.PersistentClient(path="./chroma_db")

# 2. Define the BGE-M3 embedding function
bge_m3_ef = SentenceTransformerEmbeddingFunction(
    model_name="BAAI/bge-m3",
    device="cpu",  # Change to "cpu" if you do not have a GPU
    normalize_embeddings=True # Highly recommended by BAAI for Cosine Similarity
)

# 3. Create or load collection 
# Note: Using 'cosine' space is standard for normalized BGE-M3 embeddings
collection = client.get_or_create_collection(
    name="bge_m3_dense",
    embedding_function=bge_m3_ef,
    metadata={"hnsw:space": "cosine"} 
)

# 4. Add some documents
collection.add(
    documents=[
        "BGE-M3 is a multilingual, multi-functional model supporting over 100 languages.",
        "Chroma is an open-source AI-native vector database designed for developer productivity."
    ],
    ids=["doc1", "doc2"]
)

# 5. Query the collection
results = collection.query(
    query_texts=["How many languages does BGE-M3 support?"],
    n_results=1
)

print(results["documents"])
```
