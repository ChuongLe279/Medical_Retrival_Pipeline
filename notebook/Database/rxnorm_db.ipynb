```python
"""Kiểm tra xem đã tạo bảng rxnorm.db đúng như mong muốn chưa"""
import sqlite3
from pathlib import Path

PROJECT_ROOT = Path.cwd().parents[1]
DB_PATH = PROJECT_ROOT / "data" / "db" / "RxNorm" / "rxnorm.db"


conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()
```

```python
"""Đầu tiên là kiểm tra có table nào"""
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
table = cursor.fetchall()
print(f"Table name = {table}")
if len(table) == 3:
    print("Đúng 3 bảng mong muốn")
```

```python
"""Kiểm tra tên cột"""
try:
    with sqlite3.connect(DB_PATH) as conn:
        cur = conn.cursor()
        cur.execute("PRAGMA table_info(RXNCONSO)")
        columns = [row[1] for row in cur.fetchall()]
        print(columns)
        print(len(columns))
        cursor.execute('select count(*) from RXNREL')
        row_count = cursor.fetchone()[0]
        print(f"Total rows = {row_count:,}")
except sqlite3.Error as e:
    print(e)
```
